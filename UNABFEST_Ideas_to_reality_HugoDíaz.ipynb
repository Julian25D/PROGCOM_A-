# Ideas to reality: Diálogos para la acción

#¿Qué es la transdisciplinaridad?
La transdisciplinaridad es un enfoque que busca responder preguntas complejas y socialmente relevantes mediante la integración profunda de diferentes disciplinas. Va más allá de la simple colaboración entre saberes: busca una transformación del pensamiento desde la raíz.

#Tipos de enfoques transdisciplinarios
* TD Trascendental: Basado en la "lógica del tercero incluido", permite percibir otros niveles de realidad donde las contradicciones aparentes pueden coexistir de manera complementaria.

* TD enfocada en soluciones: Se centra en enfrentar la complejidad de los problemas reales con respuestas concretas que integran múltiples saberes.

* TD como un "tercer espacio": Espacio de encuentro entre saberes académicos y comunitarios. Por ejemplo, los estudios étnicos vistos desde una mirada decolonial y transdisciplinaria.

# Marco de investigación basado en el deseo

Según Eve Tuck, este marco se enfoca en comprender la complejidad, contradicción y autodeterminación de las vidas vividas. No se trata solo de identificar daños, sino de escuchar profundamente los deseos de la comunidad.

Un ejemplo se dio en la isla de Vieques, donde, aunque un proyecto no trataba directamente la salud mental, se reconoció su importancia al escuchar las preocupaciones reales de la comunidad.

# oolbox Dialogue: un método para la integración

Este método promueve el diálogo profundo entre miembros de equipos diversos (científicos, académicos, comunidad). A través de encuestas con afirmaciones clave (por ejemplo, “Los deseos de la comunidad son primordiales en una investigación”), los participantes comparan sus respuestas y dialogan sobre sus diferencias. Este proceso ayuda a:

* Visibilizar valores individuales y colectivos.

* Crear entendimientos compartidos.

* Integrar conocimientos más allá del acuerdo superficial.

#Tres puntos clave de la charla

* La transdisciplinaridad no es solo sumar disciplinas, sino crear espacios donde los saberes se transforman mutuamente para abordar problemas complejos.

* Escuchar los deseos reales de la comunidad y reconocer la contradicción y la complejidad es fundamental para generar conocimiento relevante y transformador.

* Metodologías como el Toolbox Dialogue permiten construir entendimientos colectivos y éticos en equipos transdisciplinarios, a través del diálogo genuino y continuo

# Globalización y educación

Tatiana Gelvez y María Francisca Cepeda reflexionan sobre cómo fortalecer el vínculo entre la educación superior, la investigación, y las necesidades reales de los territorios desde una perspectiva internacional e inclusiva.

# Infraestructura, investigación y desarrollo territorial

Existe un gran potencial en las regiones para diversificar la economía más allá de la dependencia actual. Aspectos como el turismo, la discapacidad, y la fisioterapia se pueden integrar con proyectos de investigación universitaria, generando impacto en la vida cotidiana de las comunidades.
Esto exige un currículum más globalizado, que conecte lo local con metodologías y experiencias internacionales.

# Currículum globalizado e internacionalización del aprendizaje
Se busca enriquecer lo local con referentes globales, generando un ida y vuelta de conocimiento. Se promueven metodologías como el aprendizaje colaborativo internacional en línea (COIL), que permite a estudiantes trabajar juntos desde distintos países.

El desarrollo de habilidades blandas y competencias interculturales es crucial para la formación integral de los estudiantes, También se destaca la importancia de:

* Internacionalizar el currículum y hacer que los programas educativos sean atractivos internacionalmente.

* Fomentar el aprendizaje de idiomas para facilitar la interacción global.

* Aprovechar escuelas de verano en inglés como oportunidades para abrir el campo internacional a más estudiantes.

# Movilidad, cooperación e identidad internacional

Tatiana Gelvez subraya la importancia de los indicadores de movilidad entrante, como los estudiantes extranjeros que vienen a Colombia. Este tipo de intercambio ayuda a posicionar a Colombia como un lugar donde se puede aprender con calidad.

Además, se propone:

* Promover la cooperación internacional en investigación.

* Reconocer el valor del talento humano colombiano a nivel global como ventaja competitiva.

* Creer en la capacidad del país para ser actor válido en las conversaciones internacionales, sin arrogancia, pero con seguridad.

* Aprovechar los lazos con regiones como Europa y China para potenciar oportunidades.

# Tres puntos clave de la charla

* Una educación globalizada fortalece la conexión entre investigación, comunidad y desarrollo económico local, permitiendo que lo global enriquezca lo local y viceversa.

* La internacionalización del currículum, mediante estrategias como el aprendizaje colaborativo en línea y programas en inglés, impulsa competencias clave en los estudiantes.

* Colombia tiene el potencial de ser un actor fuerte en el escenario educativo internacional, tanto por su talento humano como por sus iniciativas académicas abiertas a la cooperación global.

# Modelo de Ciencia Abierta de la UNESCO

Charla dada por el Dr. Efrén Moreno

# ¿Qué es la ciencia?

Según la UNESCO y presentado por el Dr. Efrén Moreno, la ciencia es un esfuerzo organizado y sistemático llevado a cabo por diversos actores para estudiar objetivamente los fenómenos observados, validar sus resultados y comprender las causas, relaciones o interacciones subyacentes. Su finalidad es generar conocimiento confiable y útil para la sociedad.

# Principales actores en la ciencia:

* Instituciones generadoras de conocimiento.

* Académicos, investigadores y científicos.

* Revistas científicas, libros y artículos.

* Documentos, publicaciones y conferencias.

# ¿Qué es la Ciencia Abierta?

La Ciencia Abierta, como explicó el Dr. Moreno, es un enfoque que busca que el conocimiento científico sea accesible, reutilizable y comprendido por una amplia variedad de actores. Promueve procesos de investigación transparentes y colaborativos, en los que científicos y agentes sociales participan en la creación, evaluación y difusión del conocimiento, más allá de la comunidad académica tradicional.

# Elementos clave del Modelo de Ciencia Abierta UNESCO

* Participación abierta: Acceso universal a la información científica, sin importar nacionalidad, raza, género, edad, nivel socioeconómico, idioma, región, etnia o discapacidad.

* Infraestructuras compartidas: Plataformas físicas y virtuales que facilitan la colaboración y el acceso a recursos de investigación.

* Diálogo intercultural: Fomento de la inclusión y el respeto a diversos saberes y epistemologías culturales, integrando conocimientos tradicionales y comunitarios.

* Rol activo del científico: Los investigadores actúan como facilitadores de la colaboración y el diálogo con diferentes agentes sociales.

* Importancia de los metadatos: Datos estructurados que describen recursos científicos (autoría, fecha, condiciones de uso) para garantizar la trazabilidad y reutilización del conocimiento.

# Tres puntos clave de la charla

* Acceso universal e inclusivo al conocimiento: La ciencia abierta promueve que cualquier persona, sin importar su contexto, pueda acceder y participar en la generación de conocimiento.

* Colaboración entre ciencia y sociedad: Se rompe la barrera entre academia y comunidad, permitiendo que diversos actores contribuyan con sus saberes para crear soluciones inclusivas y culturalmente diversas.

* Infraestructuras y datos abiertos: La disponibilidad de infraestructuras compartidas y metadatos facilita la cooperación, transparencia y expansión del conocimiento científico.
